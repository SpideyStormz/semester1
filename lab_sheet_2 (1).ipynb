#question 6

```python
s=int(input("enter the number of due days:"))
r=[]
for i in range(1,s+1):
    if(i>=1 and i<=7):
        r.append(2)
    elif(i>=8 and i<=14):
        r.append(3)
    elif (i>=15 and i<=30):
        r.append(5)
    elif (i<30):
        r.append(10)
a=sum(r)
print("the total fine for",s,"days is",a,"rupees")
    
    
```
